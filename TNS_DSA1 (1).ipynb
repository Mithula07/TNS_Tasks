1. Reverse an Array
Write a program to reverse the elements of an array without using a built-in reverse function.

```python
arr = [10, 20, 30, 40, 50]

reversed_arr = []

for i in range(len(arr) - 1, -1, -1):
    reversed_arr.append(arr[i])

print(reversed_arr)
```

2. Count Even, Odd and Zero
Write a program to count the number of even numbers, odd numbers, and zeros in an array.

```python
arr = [10, 5, 0, 7, 8, 0, 13, 4]

even = 0
odd = 0
zero = 0

for i in arr:
    if i == 0:
        zero += 1
    elif i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
print("Zero:", zero)
```

3. Sum of Positive and Negative Numbers
Write a program to find the separate sum of positive and negative numbers in an array.

```python
arr = [10, -5, 20, -8, 15, -2]

positive_sum = 0
negative_sum = 0

for i in arr:
    if i > 0:
        positive_sum += i
    elif i < 0:
        negative_sum += i

print("Positive Sum:", positive_sum)
print("Negative Sum:", negative_sum)
```

4. Remove Duplicate Elements
Write a program to remove duplicate elements from an array.

```python
arr = [10, 20, 10, 30, 20, 40, 30]

result = []

for i in arr:
    if i not in result:
        result.append(i)

print(result)
```

5. Find the Missing Number
An array contains numbers from 1 to N, but one number is missing. Find the missing number.

```python
arr = [1, 2, 3, 5, 6, 7]

n = 7

expected_sum = 0
actual_sum = 0

for i in range(1, n + 1):
    expected_sum += i

for i in arr:
    actual_sum += i

missing = expected_sum - actual_sum

print("Missing Number:", missing)
```

6. Rotate an Array
Write a program to rotate an array to the right by K positions.

```python
arr = [1, 2, 3, 4, 5]
k = 2

for i in range(k):
    last = arr[-1]

    for j in range(len(arr) - 1, 0, -1):
        arr[j] = arr[j - 1]

    arr[0] = last

print(arr)
```

7. Find the Most Frequent Element
Write a program to find the element that occurs the maximum number of times in an array.

```python
arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

max_frequency = 0
most_frequent = arr[0]

for i in arr:
    frequency = 0

    for j in arr:
        if i == j:
            frequency += 1

    if frequency > max_frequency:
        max_frequency = frequency
        most_frequent = i

print("Most Frequent Element:", most_frequent)
print("Frequency:", max_frequency)
```

8. Maximum Subarray Sum
Write a program to find the maximum sum of a contiguous subarray.

```python
arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

current_sum = arr[0]
max_sum = arr[0]

start = 0
end = 0
temp_start = 0

for i in range(1, len(arr)):

    if arr[i] > current_sum + arr[i]:
        current_sum = arr[i]
        temp_start = i
    else:
        current_sum = current_sum + arr[i]

    if current_sum > max_sum:
        max_sum = current_sum
        start = temp_start
        end = i

print("Maximum Subarray Sum:", max_sum)

print("Subarray:")
print(arr[start:end + 1])
```

9. Check Whether Two Arrays are Equal
Given two arrays, check whether they contain the same elements with the same frequency,
regardless of their order.

```python
arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

if len(arr1) != len(arr2):
    print("Arrays are Not Equal")
else:

    equal = True

    for i in arr1:

        count1 = 0
        count2 = 0

        for j in arr1:
            if i == j:
                count1 += 1

        for j in arr2:
            if i == j:
                count2 += 1

        if count1 != count2:
            equal = False
            break

    if equal:
        print("Arrays are Equal")
    else:
        print("Arrays are Not Equal")
```
