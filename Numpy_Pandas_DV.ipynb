Problem 1: Student Marks Analysis

```python
import numpy as np

# Create a NumPy array containing marks of 10 students
marks = np.array([85, 72, 91, 68, 77, 95, 64, 88, 73, 80])

# Display the marks
print("Student Marks:", marks)

# Find total marks
total = np.sum(marks)
print("Total Marks:", total)

# Find average marks
average = np.mean(marks)
print("Average Marks:", average)

# Find highest marks
highest = np.max(marks)
print("Highest Marks:", highest)

# Find lowest marks
lowest = np.min(marks)
print("Lowest Marks:", lowest)

# Display marks greater than 75
greater_than_75 = marks[marks > 75]
print("Marks greater than 75:", greater_than_75)
```

Problem 2: Temperature Analysis

```python
import numpy as np

# Create a NumPy array containing temperatures for 7 days
temperatures = np.array([28, 31, 29, 33, 35, 30, 32])

# Display temperatures
print("Temperatures:", temperatures)

# Find average temperature
average = np.mean(temperatures)
print("Average Temperature:", average)

# Find highest temperature
highest = np.max(temperatures)
print("Highest Temperature:", highest)

# Find lowest temperature
lowest = np.min(temperatures)
print("Lowest Temperature:", lowest)

# Display days where temperature was above 30°C
above_30 = temperatures[temperatures > 30]
print("Temperatures above 30°C:", above_30)

# Increase all temperatures by 2°C
updated_temperatures = temperatures + 2
print("Updated Temperatures:", updated_temperatures)
```

Problem 3: Student Performance DataFrame

```python
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["Arun", "Priya", "Rahul", "Divya", "Karthik", "Sneha", "Vijay", "Anu"],
    "Department": ["AI&DS", "CSE", "ECE", "AI&DS", "CSE", "AI&DS", "ECE", "CSE"],
    "Marks": [85, 72, 91, 68, 77, 95, 64, 80],
    "Attendance": [90, 75, 88, 82, 78, 92, 70, 85]
}

df = pd.DataFrame(data)

# Display the complete DataFrame
print("Student Data:")
print(df)

# Display first 5 students
print("\nFirst 5 Students:")
print(df.head(5))

# Find average marks
average_marks = df["Marks"].mean()
print("\nAverage Marks:", average_marks)

# Display students who scored more than 75
print("\nStudents who scored more than 75:")
print(df[df["Marks"] > 75])

# Display students whose attendance is below 80%
print("\nStudents with attendance below 80%:")
print(df[df["Attendance"] < 80])

# Sort students based on marks
print("\nStudents sorted by Marks:")
print(df.sort_values("Marks"))
```

Problem 4: Product Sales Analysis

```python
import pandas as pd

# Create a DataFrame
data = {
    "Product Name": ["Laptop", "Mobile", "Headphones", "Keyboard", "Monitor", "Mouse"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Electronics", "Accessories"],
    "Price": [60000, 25000, 2000, 1500, 12000, 800],
    "Quantity Sold": [20, 75, 120, 60, 35, 100]
}

df = pd.DataFrame(data)

# Display the DataFrame
print("Product Data:")
print(df)

# Calculate total sales for each product
df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("\nTotal Sales for Each Product:")
print(df)

# Find the product with the highest sales
highest_sales_product = df.loc[df["Total Sales"].idxmax()]

print("\nProduct with Highest Sales:")
print(highest_sales_product)

# Find average product price
average_price = df["Price"].mean()

print("\nAverage Product Price:", average_price)

# Display products where quantity sold is greater than 50
print("\nProducts with Quantity Sold greater than 50:")
print(df[df["Quantity Sold"] > 50])

# Sort products based on total sales
print("\nProducts Sorted by Total Sales:")
print(df.sort_values("Total Sales"))
```

Problem 5: Monthly Sales Visualization

```python
import matplotlib.pyplot as plt

# Dataset
months = ["January", "February", "March", "April", "May", "June"]
sales = [25000, 30000, 28000, 35000, 40000, 38000]

# -------------------------------
# Line Chart
# -------------------------------

plt.plot(months, sales, marker="o", label="Monthly Sales")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()

plt.show()


# -------------------------------
# Bar Chart
# -------------------------------

plt.bar(months, sales, label="Monthly Sales")

plt.title("Monthly Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales (₹)")
plt.legend()

plt.show()
```

Problem 6: Student Marks Visualization

```python
import matplotlib.pyplot as plt

# Dataset
students = [
    "Student 1", "Student 2", "Student 3", "Student 4", "Student 5",
    "Student 6", "Student 7", "Student 8", "Student 9", "Student 10"
]

marks = [85, 72, 91, 55, 38, 67, 82, 45, 76, 29]


# --------------------------------
# Bar Chart - Student Marks
# --------------------------------

plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")

# Rotate student names for readability
plt.xticks(rotation=45)

plt.show()


# --------------------------------
# Performance Categories
# --------------------------------

excellent = 0
good = 0
average = 0
needs_improvement = 0

for mark in marks:

    if 80 <= mark <= 100:
        excellent += 1

    elif 60 <= mark <= 79:
        good += 1

    elif 40 <= mark <= 59:
        average += 1

    else:
        needs_improvement += 1


# Store category names
categories = [
    "Excellent",
    "Good",
    "Average",
    "Needs Improvement"
]

# Store number of students in each category
counts = [
    excellent,
    good,
    average,
    needs_improvement
]


# --------------------------------
# Pie Chart
# --------------------------------

plt.pie(
    counts,
    labels=categories,
    autopct="%1.1f%%"
)

plt.title("Student Performance Distribution")

plt.show()
```
